```python
# Display first 5 rows
print("First 5 rows:")
print(df.head())
```

```python
# Q2: Mean, Median, Minimum and Maximum
print("\nQ2: Students Statistics")

print("Mean:", df["Students"].mean())
print("Median:", df["Students"].median())
print("Minimum:", df["Students"].min())
print("Maximum:", df["Students"].max())
```

```python
# Q3: Descriptive Statistics
print("\nQ3: Descriptive Statistics")
# Descriptive Statistics
descriptive = df[["Age", "Beauty", "Evaluation", "Students"]].describe()

print(descriptive)
```

```python
# Q4: Histogram of Beauty
plt.hist(df["Beauty"], bins=10)
plt.xlabel("Beauty Score")
plt.ylabel("Number of Teachers")
plt.title("Histogram of Beauty Scores")
plt.show()
```

```python

# Q5: Beauty by Gender
print("\nQ5: Beauty Score by Gender")
print(df.groupby("Gender")["Beauty"].agg(["mean", "std"]))
```

```python
# Q6: Tenure by Gender
print("\nQ6: Tenure by Gender")


table = pd.crosstab(df["Gender"], df["Tenure"])
print(table)
```

```python
# Calculate percentage
percentage = table.div(table.sum(axis=1), axis=0) * 100

print("\nPercentage of Tenured Teachers:")
print(percentage["Yes"])
```
